```python
# Image Quality Metrics - SSIM and PSNR Calculator
import numpy as np
import cv2
import ImageReward as RM
from PIL import Image
```

```python
model = RM.load("ImageReward-v1.0")
```

```python
def grade(newImagePath):
  image = Image.open(newImagePath)

  # prompt = "A photorealistic image of a red ball on a white surface. The ball casts a dark shadow. The background is a blue sky."
  # prompt = "A photorealistic image of a stone ball on a wooden table. The ball is casting a dark shadow. The background is a sky out of focus with clouds."
  # prompt = "A photorealistic image of a tennis ball on a tree stump. The ball is casting a dark shadow. The background is a rainforest out of focus"
  # prompt = "A photorealistic image of a red rubber ball being reflected in a brown mirror on a white surface. The mirror also reflects the small shadow underneath the ball. The ball casts a dark shadow. The background is a blue sky."
  # prompt = "A photorealistic image of a grey rock ball being reflected in an oak mirror on the beach. The mirror also reflects the small round shadow underneath the ball. The ball casts a dark shadow. The background is a blue sky with clouds."
  # prompt = "A photorealistic image of a basketball being reflected in a futuristic mirror on mars. The mirror also reflects the small round shadow underneath the basketball. The basketball casts a dark shadow. The background is mars hills, dust-storms and space out of focus."

  # prompt = "A photorealistic image of a green glass with a red straw on a white surface. The glass is casting a dark shadow. The background is a blue sky. The straw and background are refracted in the glass."
  # prompt = "A photorealistic image of an engraved glass cup with a metal straw on an white table. The cup is casting a dark shadow on the table. The background is a sky with many clouds out of focus. The straw, sky and clouds are refracted in the cup."
  prompt = "A photorealistic image of a bumpy clear empty glass with a wooden straw on a wooden table. The glass is casting a dark shadow. The background is a beautiful forest and sky of focus. The straw and forest are refracted in the glass with a clear curve."

  reward = model.score(prompt, image)
  reward
  return reward
```

```python
print(grade("original.png"))
print(grade("2.1-SD.png"))
print(grade("2.2-SCN.png"))
print(grade("2.3-MCND.png"))
print(grade("2.4-MCNN.png"))
```
